## <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">5 · JSON — 프로그램끼리 주고받는 글꼴</mark>


### 왜 필요한가

1. 5교시에 만든 딕셔너리는 **코랩을 닫으면 사라집니다.** 파일로 남겨야 다음 프로그램이 이어받습니다.
2. 그런데 딕셔너리를 그냥 파일에 쓰면 다시 읽었을 때 글자 덩어리일 뿐입니다. 약속된 글꼴이 필요합니다.
3. 그 약속이 **JSON** 입니다. 파이썬·자바·자바스크립트가 모두 읽을 수 있습니다.


### 이 시간에 나오는 말

| 말 | 뜻 |
|---|---|
| JSON | 프로그램끼리 데이터를 주고받기로 약속한 글꼴 |
| `dumps` | 파이썬 값 → **글자** (s 는 string) |
| `loads` | 글자 → 파이썬 값 |
| `dump` | 파이썬 값 → **파일** |
| `load` | 파일 → 파이썬 값 |
| `ensure_ascii` | 한글을 그대로 쓸지 정하는 설정 |
| `indent` | 줄을 나눠 보기 좋게 할지 정하는 설정 |


### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">6.1 `dumps` 와 `loads` — 글자로 바꾼다</mark>

```python
import json

menu = {"name": "coffee", "price": 4500}

text = json.dumps(menu)     # 딕셔너리 → 글자
print(text)                 # {"name": "coffee", "price": 4500}

back = json.loads(text)     # 글자 → 딕셔너리
print(back["price"])        # 4500
```

`dumps` 가 돌려주는 것은 **글자**입니다. 따옴표가 큰따옴표로 바뀌는 것도 JSON 의 약속입니다.

`sample_logs_raw.csv` 를 한 줄씩 읽어 **정규화한 딕셔너리를 리스트에 모으고** 건수를 출력하시오.

1. 빈 줄과 칸이 두 개 이하인 줄은 건너뜁니다.
2. `event` 는 대문자로, IP 가 없으면 `unknown` 으로 넣습니다.
3. 시각이 숫자로 안 바뀌는 줄은 `try`·`except ValueError` 로 건너뜁니다.
4. **건너뛴 줄도 셉니다.** 버린 것을 세지 못하면 왜 9건인지 답할 수 없습니다.

| | |
|---|---|
| 🎯 나와야 하는 결과 | `정규화 9건 · 버린 줄 2건` |

**💡 힌트**

1. 오전에 쓴 파일 읽기 뼈대를 그대로 씁니다.
2. 칸 수가 모자란 줄은 `if len(parts) < 3:` 일 때 `continue` 대신 `else` 로 감싸 건너뜁니다.
3. 딕셔너리를 만드는 줄을 `try` 안에 둡니다.


````python
# 여기에 코드를 입력하세요
logs = []
skipped = 0

with open("sample_logs_raw.csv", encoding="utf-8") as f:
    for line in f:
        parts = line.strip().split(",")

        if len(parts) < 3:
            skipped = skipped + 1
        else:
            if len(parts) >= 4:
                ip = parts[3]
            else:
                ip = "unknown"

            try:
                log = {
                    "time": parts[0],
                    "user": parts[1],
                    "event": parts[2].upper(),
                    "ip": ip,
                    "hour": int(parts[0].split(":")[0])
                }

                logs.append(log)

            except ValueError:
                skipped = skipped + 1

print(f"정규화 {len(logs)}건 · 버린 줄 {skipped}건")
````

<table width="100%" bgcolor="#A5D6A7"><tr><td bgcolor="#A5D6A7"><h3><font color="#1B5E20">✍️ 문제 7-7 · 필수 필드 확인하기</font></h3></td></tr></table>

정규화한 기록에 **없으면 안 되는 칸**이 다 들어 있는지 확인하시오.

1. 필수 칸 이름을 리스트로 맨 위에 적어 둡니다 — `time`·`user`·`event`·`ip`.
2. 기록 하나를 놓고, 필수 칸을 하나씩 돌면서 **없는 것을 모읍니다.**
3. 모은 것을 출력합니다.

| | |
|---|---|
| 주어지는 값 | `row = {"time": "09:41", "user": "guest", "event": "UNKNOWN"}` |
| 🎯 나와야 하는 결과 | `빠진 항목: ['ip']` |

**💡 힌트**

1. 없는지 보는 것은 `not in` 입니다. 지난주에 배운 `in` 의 반대입니다.
2. 빈 리스트를 하나 만들어 없는 이름을 `append` 합니다.
3. 임계값을 맨 위에 적은 것처럼, 필수 칸 목록도 맨 위에 한 번만 적습니다.


````python
row = {"time": "09:41", "user": "guest", "event": "UNKNOWN"}

REQUIRED = ["time", "user", "event", "ip"]

missing = []

for key in REQUIRED:
    if key not in row:
        missing.append(key)

print(f"빠진 항목 : {missing}")
````

2. `json.dump(logs, f, ensure_ascii=False, indent=2)` 로 저장합니다.
3. `print` 는 파일 맨 아래에 둡니다.


````python
%%writefile normalize_logs.py
import json
logs = []
skipped = 0

with open("sample_logs_raw.csv", encoding="utf-8") as f:
    for line in f:
        parts = line.strip().split(",")

        if len(parts) < 3:
            skipped = skipped + 1
        else:
            if len(parts) >= 4:
                ip = parts[3]
            else:
                ip = "unknown"

            try:
                log = {
                    "time": parts[0],
                    "user": parts[1],
                    "event": parts[2].upper(),
                    "ip": ip,
                    "hour": int(parts[0].split(":")[0])
                }

                logs.append(log)

            except ValueError:
                skipped = skipped + 1

with open("normalized_logs.json", "w", encoding="utf-8") as f:
    json.dump(logs, f, ensure_ascii=False, indent=2)

print(f"정규화 {len(logs)}건")
````

````python
#@title 정답 ⭐7-14 { display-mode: "form" }
import json

logs = []
with open("sample_logs_raw.csv", encoding="utf-8") as f:
    for line in f:
        parts = line.strip().split(",")
        if len(parts) >= 3:
            if len(parts) >= 4:
                ip = parts[3]
            else:
                ip = "unknown"
            try:
                logs.append({"time": parts[0], "user": parts[1],
                             "event": parts[2].upper(), "ip": ip,
                             "hour": int(parts[0].split(":")[0])})
            except ValueError:
                pass

report = {}
for log in logs:
    if log["event"] == "LOGIN_FAIL":
        user = log["user"]
        if user in report:
            report[user] = report[user] + 1
        else:
            report[user] = 1

with open("report.json", "w", encoding="utf-8") as f:
    json.dump(report, f, ensure_ascii=False, indent=2)

print(report)
````
